```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data.csv')
x=df.iloc[:,0]
y=df.iloc[:,1]
n=float(len(x))
plt.scatter(x,y)
plt.show()
```

```python
#Building Geadient Descent
m=0
c=0
L=0.0001
epochs=1000
n=float(len(x))
for i in range(len(x)):
    y_pred=m*x+c
    d_m=(-2/n)*sum(x*(y-y_pred))
    d_c=(-2/n)*sum(y-y_pred)
    m=m-L*d_m
    c=c-L*d_c
print(m,c)
```

```python
#prediction
y_pred=m*x+c
plt.scatter(x,y)
plt.scatter(x,y_pred)
plt.show()
```
